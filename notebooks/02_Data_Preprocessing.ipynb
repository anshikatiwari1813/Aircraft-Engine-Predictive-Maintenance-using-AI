# Aircraft Engine Predictive Maintenance

## Notebook 2: Data Preprocessing

### Objectives

- Load training dataset
- Create Remaining Useful Life (RUL)
- Drop unnecessary columns
- Scale sensor values
- Save processed dataset

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import MinMaxScaler

pd.set_option("display.max_columns", None)

print("Libraries Imported Successfully")
```

```python
column_names = [
    "engine_id",
    "cycle",
    "setting1",
    "setting2",
    "setting3"
]

for i in range(1,22):
    column_names.append(f"sensor_{i}")

train = pd.read_csv(
    "../dataset/train_FD001.txt",
    sep=r"\s+",
    header=None,
    names=column_names
)

print(train.shape)
```

```python
max_cycle = train.groupby("engine_id")["cycle"].max()

train["RUL"] = train.apply(
    lambda row: max_cycle[row["engine_id"]] - row["cycle"],
    axis=1
)

train.head()
```

```python
train.head()
```

```python
train.isnull().sum()
```

```python
constant_columns = []

for column in train.columns:
    if train[column].nunique() == 1:
        constant_columns.append(column)

constant_columns
```

```python
train = train.drop(columns=constant_columns)

print(train.shape)
```

```python
X = train.drop(columns=["RUL"])

y = train["RUL"]

print(X.shape)
print(y.shape)
```

```python
scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

X_scaled.head()
```

```python
processed_data = X_scaled.copy()

processed_data["RUL"] = y.values

processed_data.head()
```

```python
processed_data.to_csv(
    "../dataset/processed_train.csv",
    index=False
)

print("Processed Dataset Saved Successfully")
```
